```python
import pandas as pd
import numpy as np

# 1. Purana fleet data load karein
df = pd.read_csv('new healthy_fleet_data.csv')

# 2. Target: Hum sirf 'INS_Mysore' mein fault dikhayenge
target_ship = 'INS_Mysore'

# INS_Mysore ki saari rows dhoondein
mask = df['Ship_Name'] == target_ship
ship_indices = df[mask].index

# Hum sirf aakhri 200 minutes (rows) mein fault dikhayenge
fault_start = ship_indices[800] 
fault_end = ship_indices[999]

print(f"⚠️ Injecting engine fault in {target_ship} from index {fault_start} to {fault_end}...")

# 3. Increase Temperature (Gradual heating)
# INS_Mysore ka temp 450 se badhkar 510+ ho jayega
df.loc[fault_start:fault_end, 'Temperature_C'] += np.linspace(0, 60, 200)

# 4. Increase Vibration (Mechanical wear and tear)
df.loc[fault_start:fault_end, 'Vibration_mm'] += np.linspace(0, 0.06, 200)

# 5. Drop RPM slightly (Engine struggling)
df.loc[fault_start:fault_end, 'RPM'] -= np.linspace(0, 150, 200)

# 6. Maintenance Status update karein (AI Labeling)
df.loc[fault_start:fault_end, 'Maintenance_Status'] = 1

# 7. Final Training Dataset save karein
df.to_csv('new naval_fleet_with_faults.csv', index=False)

print(f"✅ Success! Faults injected only for {target_ship}. Baaki ships safe hain.")
```
