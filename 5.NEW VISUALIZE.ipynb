```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Load the data with faults
df = pd.read_csv('new naval_fleet_with_faults.csv')

# 2. Set the style for a professional look
sns.set_theme(style="darkgrid")

# 3. Create a figure with two subplots (Top for Temp, Bottom for Vibration)
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(12, 8), sharex=True)

# Plot Temperature
sns.lineplot(data=df, x=df.index, y='Temperature_C', ax=ax1, color='blue')
ax1.set_title('Naval Engine Temperature over Time')
ax1.axvline(x=800, color='red', linestyle='--', label='Fault Starts') # Mark where we injected the fault
ax1.legend()

# Plot Vibration
sns.lineplot(data=df, x=df.index, y='Vibration_mm', ax=ax2, color='orange')
ax2.set_title('Naval Engine Vibration over Time')
ax2.axvline(x=800, color='red', linestyle='--')

plt.tight_layout()
plt.show()
```
