```python
import pandas as pd
import numpy as np
from datetime import datetime, timedelta

# 1. Seed for consistency
np.random.seed(42)

# 2. Ships ki list (Aap yahan aur naam add kar sakte hain)
ships = ['INS_Vikrant', 'INS_Vikramaditya', 'INS_Mysore']
rows_per_ship = 1000  # Har ship ke liye 1000 rows

all_ship_data = []

start_time = datetime.now()

# 3. Loop: Har ship ke liye data generate karein
for ship_name in ships:
    print(f"Generating data for {ship_name}...")
    
    # Timestamps for this specific ship
    timestamps = [start_time + timedelta(minutes=i) for i in range(rows_per_ship)]
    
    # Sensors (Healthy range)
    temp = 450 + np.random.normal(0, 2, rows_per_ship)
    pressure = 30 + np.random.normal(0, 0.5, rows_per_ship)
    vibration = 0.03 + np.random.normal(0, 0.002, rows_per_ship)
    rpm = 3600 + np.random.normal(0, 10, rows_per_ship)
    
    # Create DataFrame for this ship
    ship_df = pd.DataFrame({
        'Timestamp': timestamps,
        'Ship_Name': ship_name,  # Ab yahan ship ka naam aayega
        'Engine_ID': f'GT_{ship_name[:3].upper()}_01', # Unique Engine ID
        'Temperature_C': temp,
        'Pressure_PSI': pressure,
        'Vibration_mm': vibration,
        'RPM': rpm,
        'Maintenance_Status': 0
    })
    
    all_ship_data.append(ship_df)

# 4. Saari ships ka data ek saath jodein (Concatenate)
final_fleet_data = pd.concat(all_ship_data, ignore_index=True)

# 5. Save karein
final_fleet_data.to_csv('new healthy_fleet_data.csv', index=False)

print("\n⚓ Success! 'healthy_fleet_data.csv' created with 3 ships.")
print(final_fleet_data.sample(10)) # Beech mein se 10 random rows dikhayega
```
